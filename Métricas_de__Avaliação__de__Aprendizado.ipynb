```python
def calcular_metrica(matriz_confusao):

    VP = matriz_confusao.get("VP", 0)
    VN = matriz_confusao.get("VN", 0)
    FP = matriz_confusao.get("FP", 0)
    FN = matriz_confusao.get("FN", 0)

    # Sensibilidade (Recall)
    sensibilidade = VP / (VP + FN) if (VP + FN) > 0 else 0

    # Especificidade
    especificidade = VN / (FP + VN) if (FP + VN) > 0 else 0

    # Acurácia
    total = VP + VN + FP + FN
    acuracia = (VP + VN) / total if total > 0 else 0

    # Precisão
    precisao = VP / (VP + FP) if (VP + FP) > 0 else 0

    # F-score
    f_score = (2 * precisao * sensibilidade) / (precisao + sensibilidade) if (precisao + sensibilidade) > 0 else 0


    return {
        "Sensibilidade (Recall)": sensibilidade,
        "Especificidade": especificidade,
        "Acurácia": acuracia,
        "Precisão": precisao,
        "F-score": f_score,
    }


matriz_confusao = {
    "VP": 50,  # Verdadeiros Positivos
    "VN": 100,  # Verdadeiros Negativos
    "FP": 10,  # Falsos Positivos
    "FN": 18   # Falsos Negativos
}


metricas = calcular_metrica(matriz_confusao)


import pandas as pd
df_metricas = pd.DataFrame(metricas.items(), columns=["Métrica", "Valor"])
print(df_metricas)


import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))
sns.barplot(x="Métrica", y="Valor", data=df_metricas)
plt.title("Métricas de Avaliação")
plt.ylabel("Valor")
plt.xticks(rotation=45)
plt.show()
```
